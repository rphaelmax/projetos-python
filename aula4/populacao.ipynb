```python
import pandas as pd
import openpyxl
```

```python
path = r"H:\3ª ETAPA\Python\projetos-python\aula4\CD2022_Populacao_2010_Compatibilizada_20231222.xlsx"
```

```python
tabela = pd.read_excel(path, header=2)
```

```python
print(tabela.columns)
```

```python
tabela
```

```python
col_2010 = 'População 2010 (Alterações de Limites até 2022)1'
col_2022 = 'População Censo 2022'

pop_estado = (
    tabela.groupby(["COD. UF", "UF"], as_index=False)
      .agg(
          populacao_2010=(col_2010, "sum"),
          populacao_2022=(col_2022, "sum"),
      )
)

pop_estado["crescimento"] = pop_estado["populacao_2022"] - pop_estado["populacao_2010"]
pop_estado = pop_estado.sort_values("crescimento", ascending=False).reset_index(drop=True)
pop_estado
```

```python
pop_estado.to_csv("populacao_por_estado.csv", index=False)
```

```python
pop_municipio = tabela[["COD. UF", "UF", "COD. MUNIC", "NOME DO MUNICÍPIO", col_2010, col_2022]].copy()
pop_municipio = pop_municipio.rename(columns={col_2010: "populacao_2010", col_2022: "populacao_2022"})

pop_municipio["crescimento"] = pop_municipio["populacao_2022"] - pop_municipio["populacao_2010"]
pop_municipio = pop_municipio.sort_values("crescimento", ascending=False).reset_index(drop=True)
pop_municipio
```

```python
pop_municipio.to_csv("populacao_por_municipio.csv", index=False)
```
